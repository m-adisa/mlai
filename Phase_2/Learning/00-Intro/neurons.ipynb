```python
# Basic Neuron
import numpy as np

inputs = np.array([1.0, 2.0, 3.0, 2.5])

outputs = np.array([0.5, 1.0, 1.5, 2.0])

bias = 2.0

neural_output = np.dot(inputs, outputs) + bias

print(neural_output)
```

```python
# Basic Neural Layer

layer_input = np.array([1.0, 2.0, 3.0, 2.5])

layer_weights = np.array([[0.2, 0.8, -0.5, 1.0],
                          [0.5, -0.91, 0.26, -0.5],
                          [-0.26, -0.27, 0.17, 0.87]
                        ])

layer_biases = np.array([2.0, 3.0, 0.5])

layer_output = np.dot(layer_weights, layer_input) + layer_biases # same as np.dot(layer_input, layer_weights.T) + layer_biases

print(layer_output)
```

```python
# Batch of Data

batch_inputs = np.array([[1.0, 2.0, 3.0, 2.5],
                         [2.0, 5.0, -1.0, 2.0],
                         [-1.5, 2.7, 3.3, -0.8]
                        ])

weights = np.array([[0.2, 0.8, -0.5, 1.0],
                    [0.5, -0.91, 0.26, -0.5],
                    [-0.26, -0.27, 0.17, 0.87]
                  ])

biases = np.array([2.0, 3.0, 0.5])

batch_output = np.dot(batch_inputs, weights.T) + biases # transpose weights to match the shape of batch_inputs

print(batch_output)
```

```python
# Multiple Layers

input_matrix = np.array([[1.0, 2.0, 3.0, 2.5],
                         [2.0, 5.0, -1.0, 2.0],
                         [-1.5, 2.7, 3.3, -0.8]
                        ])

weights1 = np.array([[0.2, 0.8, -0.5, 1.0],
                     [0.5, -0.91, 0.26, -0.5],
                     [-0.26, -0.27, 0.17, 0.87]
                   ])

biases1 = np.array([2.0, 3.0, 0.5])

weights2 = np.array([[0.1, -0.14, 0.5],
                     [-0.5, 0.12, -0.33],
                     [-0.44, 0.73, -0.13]
                   ])

biases2 = np.array([-1.0, 2.0, -0.5])

layer1_output = np.dot(input_matrix, weights1.T) + biases1
layer2_output = np.dot(layer1_output, weights2.T) + biases2

print(layer2_output)
```
